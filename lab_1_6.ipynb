Условие: Удаление всех соседних дубликатов в строке (напр. "abbaca" -> "ca").
4. Описание реализации: использовала встроенный тип list в роли стека; сложность -  O(n)

```python
def remove_duplicates(s: str) -> str:
    stack = []

    for char in s:    # если стек не пуст и последний элемент равен текущему символу — удаляем его из стека
        if stack and stack[-1] == char:
            stack.pop()    #снять символ с вершины стека
        else:
            stack.append(char)     #положить символ на вершину стека

    return "".join(stack)    #преобразование стека в строку

print(remove_duplicates(input()))    #проверка
```
