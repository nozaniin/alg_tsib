# Сравнение хешей

Функция check_integrity(data_str, given_hash) возвращает True, если MD5 от data_str совпадает с given_hash.

Шаблон:
def check_integrity(data_str: str, given_hash: str) -> bool:
pass

Тесты:

 Input: admin|21232f297a57a5a743894a0e4a801fc3 -> Output: True

 Input: user|21232f297a57a5a743894a0e4a801fc3 -> Output: False


```python
import hashlib   # подключаем модуль для работы с хэш-функциями

def check_integrity(data_str: str, given_hash: str) -> bool:
    """Возвращает True, если MD5 от data_str совпадает с given_hash."""

    # Вычисляем MD5-хэш от data_str:
    # 1) data_str.encode('utf-8') – переводим строку в байты (кодировка UTF-8).
    # 2) hashlib.md5(...) – создаём объект MD5 и подаём на вход байты.
    # 3) .hexdigest() – получаем хэш в виде 32-символьной шестнадцатеричной строки.
    computed = hashlib.md5(data_str.encode('utf-8')).hexdigest()
    return computed == given_hash

# тесты из задания
print(check_integrity(input(), input()))   #для первой проверки
print(check_integrity(input(), input()))   #для второй проверки
```
