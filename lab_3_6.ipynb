Реализуйте функцию bubble_sort(arr), которая сортирует список целых

список).

Тестовые данные:
a = [5, 2, 9, 1, 5, 6]; bubble_sort(a); print(a)

Ожидается: [1, 2, 5, 5, 6, 9]


```python
def bubble_sort(arr):
    """
    Сортирует список целых чисел по возрастанию методом пузырька
    """
    n = len(arr)
    for i in range(n):
        # Флаг для оптимизации: если за проход не было обменов, сортировка завершена
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break


# Проверка (тестовые данные из задания)
if __name__ == "__main__":
  a = [5, 2, 9, 1, 5, 6]
  bubble_sort(a)
  print(a)   # ожидается: [1, 2, 5, 5, 6, 9]
```
